```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('terror/arms_DF_ver2.csv', encoding='ISO-8859-1')
```

```python
df = df.sort_values(["weapontype"], ascending=[True])
df
```

```python
weapontype_num = df.groupby('weapontype')
weapontype_num.size()
```

```python
df
```

```python
point_data = df[['lat','lon']]
point_data
```

```python
point_target = [0]*203 +[1]*77755 +[2]*49801 +[3]*2922
point_target
```

```python
from sklearn.model_selection import train_test_split
```

```python
#test_size: Decide how many percentages to allocate
#random_state= random seed(to make it random)
x_train, x_test, y_train, y_test = train_test_split(point_data, point_target, shuffle=True, test_size=0.2, random_state=20)
```

```python
print(len(x_train))
print(len(x_test))
print(len(y_train))
print(len(y_test))
```

```python
#Load the knn library from sklearn
from sklearn.neighbors import KNeighborsClassifier
```

```python
kn = KNeighborsClassifier()
```

```python
#Fit the train data
kn.fit(x_train,y_train)
```

```python
kn.score(x_train,y_train)
```

```python
kn.score(x_test,y_test)
```

```python
#Let's make a new model to test
testpoint_D = 15.675051    
testpoint_W = 120.331618

#Put the latitude and longitude values into the KNN algorithm.
testpoint = kn.predict([[testpoint_D,testpoint_W]])

#Visualization step (scatter plot)
plt.scatter(df['lat'],df['lon'])
plt.scatter(testpoint_D,testpoint_W,marker='s')

plt.xlabel('lat')
plt.ylabel('lon')

plt.title("GTD Normal")
plt.show()


#print kn.predict value
print(testpoint)
```

```python
#Matches a string that matches each number
if testpoint == 0:
  print("Chemical")
elif testpoint == 1:
  print("explosives/Bombs/Dynamite")
elif testpoint == 2:
  print("Firearms")
elif testpoint == 3:
  print("Melee")
else: print("error")
```

```python
C_lat = df.loc[df['weapontype'] == 'Chemical', 'lat'] 
C_lon = df.loc[df['weapontype'] == 'Chemical', 'lon'] 
E_lat = df.loc[df['weapontype'] == 'Explosives/Bombs/Dynamite', 'lat'] 
E_lon = df.loc[df['weapontype'] == 'Explosives/Bombs/Dynamite', 'lon'] 
A_lat = df.loc[df['weapontype'] == 'Firearms', 'lat'] 
A_lon = df.loc[df['weapontype'] == 'Firearms', 'lon'] 
M_lat = df.loc[df['weapontype'] == 'Melee', 'lat'] 
M_lon = df.loc[df['weapontype'] == 'Melee', 'lon'] 
```

```python
#Chemical
plt.scatter(C_lat[:],C_lon[:], c="#f39451",alpha=0.4 ,marker='x')
#Explosives/Bombs/Dynamite
plt.scatter(E_lat[:],E_lon[:], c="#66c0c8",alpha=0.4 ,marker='x')
#Firearms
plt.scatter(A_lat[:],A_lon[:], c="#fda898",alpha=0.3 ,marker='x')
#Melee
plt.scatter(M_lat[:],M_lon[:], c="yellow",alpha=0.2 ,marker='x')
#Prediction
plt.scatter(testpoint_D,testpoint_W,marker='v', c='#bb1c2a', s=250, edgecolors='black' ,linewidth=2)

plt.xlabel('lat')
plt.ylabel('lon')
plt.title("GTD Color")
ax = plt.gca()

ax.set_facecolor('#424242')

plt.show()
```
